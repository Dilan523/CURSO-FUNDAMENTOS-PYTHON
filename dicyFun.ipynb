Leer una frase y devolver el conjunto de caracteres diferentes que se encuentran presentes en la frase leída.

```python
frase = input("ingrese una frase: ")
caracteres_diferentes = set(frase)
for c in frase:
  caracteres_diferentes.add(c)
print("caracteres diferentes", caracteres_diferentes)
```

En una escuela se permite la inscripción de sus alumnos a las siguientes actividades deportivas: futbol (f), baloncesto (b). Realice un programa que permita ingresar el nombre de sus estudiantes e inscribir a cada uno en uno o los dos deportes.

```python
futbol = set()
baloncesto = set()

def menu():
    print("\n--- Menú de Opciones ---")
    print("1. Inscribir estudiante")
    print("2. Mostrar estudiantes inscritos en fútbol")
    print("3. Mostrar estudiantes inscritos en baloncesto")
    print("4. Mostrar estudiantes inscritos en ambos deportes")
    print("5. Mostrar todos los estudiantes inscritos")
    print("6. Mostrar estudiantes inscritos en solo un deporte")
    print("7. Salir")
    return input("Seleccione una opción (1-7): ")

while True:
    opcion = menu()

    if opcion == "1":
        nombre = input("Ingrese el nombre del estudiante: ")
        deporte = input("¿A qué deporte desea inscribirse? (f = fútbol, b = baloncesto, fb = ambos): ")

        if deporte == "f":
            futbol.add(nombre)
        elif deporte == "b":
            baloncesto.add(nombre)
        elif deporte == "fb":
            futbol.add(nombre)
            baloncesto.add(nombre)
        else:
            print("Opción de deporte inválida. No se realizó la inscripción.")

    elif opcion == "2":
        print("\nEstudiantes inscritos en fútbol:")
        print(futbol)

    elif opcion == "3":
        print("\nEstudiantes inscritos en baloncesto:")
        print(baloncesto)

    elif opcion == "4":
        ambos = futbol.intersection(baloncesto)
        print("\nEstudiantes inscritos en ambos deportes:")
        print(ambos)

    elif opcion == "5":
        todos = futbol.union(baloncesto)
        print("\nTodos los estudiantes inscritos en algún deporte:")
        print(todos)

    elif opcion == "6":
        solo_un_deporte = futbol.symmetric_difference(baloncesto)
        print("\nEstudiantes inscritos en solo un deporte:")
        print(solo_un_deporte)

    elif opcion == "7":
        print("Saliendo del programa.")
        break

    else:
        print("Opción inválida. Intente de nuevo.")
```

Realice un programa traductor de frutas del español al inglés. Para ello, solicite el nombre de una fruta al usuario y muestre su traducción al inglés. Si la fruta no se encuentra en el diccionario, permítale al usuario agregarla con su respectiva traducción. El programa solo termina cuando el usuario ingresa una fruta que no está en el diccionario y no desea agregarla.

```python
frutas = {"manzana": "apple", "plátano": "banana", "naranja": "orange"}

while True:
    fruta = input("Ingrese el nombre de una fruta en español: ")

    if fruta in frutas:
        print(f"La traducción de '{fruta}' es: {frutas[fruta]}")
    else:
        print(f"La fruta '{fruta}' no está en el diccionario.")
        agregar = input("¿Desea agregarla? (si/no): ")

        if agregar == "si":
            traduccion = input(f"Ingrese la traducción de '{fruta}' al inglés: ")
            frutas[fruta] = traduccion
            print(f"'{fruta}' ha sido agregada al diccionario.")
        else:
            print("Gracias por usar el traductor.")
            break
```

Realiza una función que tome una lista de números enteros y devuelva dos listas ordenadas. La primera con los números pares y la segunda con los números impares.

```python
def separar_pares_impares(numeros):
    pares = []
    impares = []
    for num in numeros:
        if num % 2 == 0:
            pares.append(num)
        else:
            impares.append(num)
    return pares, impares
lista = [4, 7, 2, 9, 1, 6, 8]
pares, impares = separar_pares_impares(lista)
print("Pares:", pares)
print("Impares:", impares)
```

Realiza una función que reciba una palabra y devuelva un diccionario donde las claves sean las diferentes letras que se encuentran en la palabra y los valores sean el número de veces que se repite cada letra en la palabra

```python
def contar_letras(palabra):
    contador = {}
    for letra in palabra:
        if letra in contador:
            contador[letra] += 1
        else:
            contador[letra] = 1
    return contador
palabra_usuario = input("Ingresa una palabra: ")
resultado = contar_letras(palabra_usuario)
print("Frecuencia de letras:", resultado)
```

```python
def numero_a_palabras(numero):
    nombres = {
        '0': 'cero',
        '1': 'uno',
        '2': 'dos',
        '3': 'tres',
        '4': 'cuatro',
        '5': 'cinco',
        '6': 'seis',
        '7': 'siete',
        '8': 'ocho',
        '9': 'nueve'
    }

    numero_str = str(numero)
    palabras = [nombres[digito] for digito in numero_str]
    return ' - '.join(palabras)
numero_usuario = input("Ingresa un número: ")
print(numero_a_palabras(numero_usuario))
```
